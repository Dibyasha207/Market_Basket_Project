```python
import pandas as pd

rules = pd.read_csv("../dataset/association_rules.csv")
```

```python
def recommend_products(product_name):

    recommendations = rules[
        rules['antecedents'].str.contains(product_name, na=False)
    ]

    recommendations = recommendations.sort_values(
        by='lift',
        ascending=False
    )

    return recommendations[
        ['consequents', 'confidence', 'lift']
    ].head(10)
```

```python
def recommend_products(product_name):

    recommendations = rules[
        rules['antecedents'].astype(str).str.contains(
            product_name,
            case=False,
            na=False
        )
    ]

    recommendations = recommendations.sort_values(
        by='lift',
        ascending=False
    )

    for _, row in recommendations.head(5).iterrows():

        print("Recommended Product:")
        print(list(eval(row['consequents'])))

        print("Confidence:", round(row['confidence']*100,2), "%")

        print("Lift:", round(row['lift'],2))

        print("------------------")
```

```python
recommend_products(
    "ALARM CLOCK BAKELIKE RED"
)
```
